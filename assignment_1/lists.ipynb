```python
# Part 1 – Lists

# Create the list with grades
grades = [15, 18, 12, 20, 16]

# 1. Print the original list
print("Original list:", grades)

# 2. Add grade 17 using append()
grades.append(17)

# 3. Sort grades from lowest to highest using sort()
grades.sort()

# 4. Show highest, lowest and total
print("Highest grade:", max(grades))
print("Lowest grade:", min(grades))
print("Total number of grades:", len(grades))

# 5. Print the final list
print("Final list:", grades)
```
