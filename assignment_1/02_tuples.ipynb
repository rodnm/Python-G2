```python
# Part 2 - Tuples

# Create the tuple
ages = (20, 25, 22, 30, 28)
```

```python
# 1. Print the tuple
print("tuple:", ages)
```

```python
# 2. First and last element
print("First element:", ages[0])
print("Last element:", ages[-1])
```

```python
# 3. Maximum, minimum and number of elements
print("Maximum value:", max(ages))
print("Minimum value:", min(ages))
print("Number of elements:", len(ages))
```

```python
# 4. Try to modify one of the elements of the tuple
# ages[0] = 99
# TypeError: 'tuple' object does not support item assignment.
# In that sense, tuples are ordered and immutable objects,
# meaning once they are created, their values cannot be modified.
```
