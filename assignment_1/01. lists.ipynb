```python
# Parte 1 – Listas

grades = [15, 18, 12, 20, 16]

# 1. Imprimir la lista original
print("Lista original:", grades)

# 2. Agregar 17 usando append()
grades.append(17)

# 3. Ordenar de menor a mayor
grades.sort()

# 4. Mostrar máximo, mínimo y total
print("Calificación más alta:", max(grades))
print("Calificación más baja:", min(grades))
print("Número total de calificaciones:", len(grades))

# 5. Imprimir lista final
print("Lista final:", grades)
```
